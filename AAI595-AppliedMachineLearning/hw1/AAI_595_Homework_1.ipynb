## Question 4
For the previous generated data, consider each row of the array as an individual data point, with the two columns representing its distinct features. Identify the data points with the smallest and largest sum of features. Mark them in red and blue in the figure respectively.

Hint: You can compute the sum of each row and store it in an array named "sums". Then find the minimum or maximum index of this array.

```python
sums = np.sum(data, axis=1)
max_index = np.argmax(sums)
min_index = np.argmin(sums)
plt.scatter(data[:, 0], data[:, 1], color='black', label='Data Point', marker='o')
plt.scatter(data[max_index, 0], data[max_index, 1], label='Largest Sum', color='blue', marker='o')
plt.scatter(data[min_index, 0], data[min_index, 1], label='Smallest Sum', color='red', marker='o')
plt.title('Data Point with Extremes Marked')
plt.legend()
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('seaborn-v0_8-whitegrid')
noise_scale = 900
number_of_samples = 70
noise_scale += stevens_id % 70
CWID = stevens_id % 4 + 3
x = 30 * (np.random.rand(number_of_samples, 1) - 0.5)
if CWID == 3:
  y = 6 * x + 7 * x**2 + 6 * x**3 + 10 * noise_scale*np.random.randn(number_of_samples, 1)
elif CWID == 4:
  y = 6 * x + 7 * x**2 + 5 * x**4 + 100 * noise_scale*np.random.randn(number_of_samples, 1)
elif CWID == 5:
  y = 6 * x + 8 * x**2 + 4 * x** 5 + 1000 * noise_scale*np.random.randn(number_of_samples, 1)
else:
  y = 6 * x + 7 * x**2 + 3 * x**6 + 10000 * noise_scale*np.random.randn(number_of_samples, 1)
plt.plot(x,y,'ro')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Noisy Data Points")
plt.legend()
plt.show()
```

```python
m_values = [1, 2, 3, 4, 5, 6]
plt.figure(figsize=(10, 8))
x_sorted = np.sort(x, axis=0)
for m in m_values:
    coeffs = np.polyfit(x.flatten(), y.flatten(), m) # Fit polynomial of degree m
    poly = np.poly1d(coeffs)
    y_pred = poly(x_sorted)
    plt.plot(x_sorted, y_pred, label=f"Degree {m}")

plt.scatter(x, y, color='red', label="Noisy Data", alpha=0.6)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Polynomial Fits for Different Degrees")
plt.legend()
plt.show()
```

```python
m_values = [1,2,3,4,5,6,7,8,9]
mse_values = []

for m in m_values:
    coeffs = np.polyfit(x.flatten(), y.flatten(), m)
    poly = np.poly1d(coeffs)
    y_pred = poly(x)
    mse = np.mean((y - y_pred) ** 2)
    mse_values.append(mse)

# MSE vs. Polynomial Degree
plt.figure(figsize=(8, 6))
plt.plot(m_values, mse_values, marker='o', linestyle='-')
plt.xlabel("Polynomial Degree")
plt.ylabel("MSE")
plt.title("MSE vs. Polynomial Degree")
plt.grid()
plt.show()

# Find best degree m (minimum MSE)
best_m = m_values[np.argmin(mse_values)]
print("Best polynomial degree based on MSE: ", best_m)
```
