# **Question 1: Linear Regression Modification**

## **Part a: Load the dataset**

```python
df = pd.read_csv('drive/MyDrive/Realestate.csv')
print("Dataset shape:", df.shape)
print(df.head())
```

## **Part b: Standardize the data**

```python
X = df.iloc[:, 1:-1]
y = df.iloc[:, -1]
scaler = StandardScaler()
# Standardize features
X_std = scaler.fit_transform(X)
# Standardize target
y_std = (y - y.mean()) / y.std()
print("Standardized features mean ≈ 0:")
print(np.mean(X_std, axis=0))
print("Standardized features std ≈ 1:")
print(np.std(X_std, axis=0))
```

## **Part c: Scikit-learn Linear Regression**

```python
# Creating and fiting model
model_sklearn = LinearRegression()
model_sklearn.fit(X_std, y_std)
# Making predictions
y_pred_sklearn = model_sklearn.predict(X_std)
# RMSE
rmse_sklearn = np.sqrt(mean_squared_error(y_std, y_pred_sklearn))
print("Scikit-learn RMSE: ", round(rmse_sklearn,4))
```

## **Part d: NumPy Linear Regression**

```python
class Linear_regression():
    def __init__(self, learning_rate, iterations):
        self.learning_rate = learning_rate
        self.iterations = iterations

    def fit(self, X, y):
        self.m, self.n = X.shape
        self.W = np.zeros(self.n)
        self.b = 0
        self.X = X
        self.y = y
        for i in range(self.iterations):
            self.update_weights()
        return self

    def update_weights(self):
        y_pred = self.predict(self.X)
        # Calculating gradients
        dW = -(2 * (self.X.T).dot(self.y - y_pred)) / self.m
        db = -2 * np.sum(self.y - y_pred) / self.m
        # Update weights
        self.W = self.W - self.learning_rate * dW
        self.b = self.b - self.learning_rate * db
        return self

    def predict(self, X):
        return X.dot(self.W) + self.b

# Train model with parameters
model_numpy = Linear_regression(learning_rate=0.01, iterations=1000)
model_numpy.fit(X_std, y_std)
y_pred_numpy = model_numpy.predict(X_std)
rmse_numpy = np.sqrt(mean_squared_error(y_std, y_pred_numpy))
print("NumPy Linear Regression RMSE: ", round(rmse_numpy, 4))
```

## **Part e: Stochastic Gradient Descent**

```python
class SGD_Linear_regression():
    def __init__(self, learning_rate, iterations):
        self.learning_rate = learning_rate
        self.iterations = iterations

    def fit(self, X, y):
        self.m, self.n = X.shape
        self.W = np.zeros(self.n)
        self.b = 0
        for i in range(self.iterations):
            # Data Shuffling
            indices = np.random.permutation(self.m)
            X_shuffled = X[indices]
            y_shuffled = y[indices]
            # Updating weights
            for j in range(self.m):
                X_j = X_shuffled[j:j+1]
                y_j = y_shuffled[j:j+1]
                y_pred = self.predict(X_j)
                # Calculating gradients
                dW = -(2 * (X_j.T).dot(y_j - y_pred))
                db = -2 * np.sum(y_j - y_pred)
                # Update weights
                self.W = self.W - self.learning_rate * dW
                self.b = self.b - self.learning_rate * db
        return self

    def predict(self, X):
        return X.dot(self.W) + self.b

# Train model with SGD
model_sgd = SGD_Linear_regression(learning_rate=0.001, iterations=50)
model_sgd.fit(X_std, y_std)
y_pred_sgd = model_sgd.predict(X_std)
rmse_sgd = np.sqrt(mean_squared_error(y_std, y_pred_sgd))
print("Stochastic Gradient Descent RMSE: ", round(rmse_sgd, 4))
```

# **Question 2: Bayesian Theorem**

## **Load data**

```python
df_bayes = pd.read_csv('drive/MyDrive/data.csv')
```

## **Part b: Calculate prior probabilities**

```python
total_samples = len(df_bayes)
p_x_0 = len(df_bayes[df_bayes['X'] == 0]) / total_samples
p_x_1 = len(df_bayes[df_bayes['X'] == 1]) / total_samples
p_y_0 = len(df_bayes[df_bayes['Y'] == 0]) / total_samples
p_y_1 = len(df_bayes[df_bayes['Y'] == 1]) / total_samples

print("p(x=0) = ", p_x_0)
print("p(x=1) = ", p_x_1)
print("p(y=0) = ", p_y_0)
print("p(y=1) = ", p_y_1)
```

## **Part c: Calculate likelihood**

```python
# p(x=0|y=0)
p_x0_y0 = len(df_bayes[(df_bayes['X'] == 0) & (df_bayes['Y'] == 0)]) / len(df_bayes[df_bayes['Y'] == 0])
# p(x=1|y=0)
p_x1_y0 = len(df_bayes[(df_bayes['X'] == 1) & (df_bayes['Y'] == 0)]) / len(df_bayes[df_bayes['Y'] == 0])
#p(x=0|y=1)
p_x0_y1 = len(df_bayes[(df_bayes['X'] == 0) & (df_bayes['Y'] == 1)]) / len(df_bayes[df_bayes['Y'] == 1])
# p(x=1|y=1)
p_x1_y1 = len(df_bayes[(df_bayes['X'] == 1) & (df_bayes['Y'] == 1)]) / len(df_bayes[df_bayes['Y'] == 1])

print("p(x=0|y=0) = ", p_x0_y0)
print("p(x=1|y=0) = ", p_x1_y0)
print("p(x=0|y=1) = ", p_x0_y1)
print("p(x=1|y=1) = ", p_x1_y1)
```

## **Part d: Calculate posterior probabilities**

```python
# P(y|x) = P(x|y)P(y)/P(x)
def calculate_posterior(p_x_y, p_y, p_x):
    return (p_x_y * p_y) / p_x

# p(y=1|x=0)
p_y1_x0 = calculate_posterior(p_x0_y1, p_y_1, p_x_0)
# p(y=0|x=0)
p_y0_x0 = calculate_posterior(p_x0_y0, p_y_0, p_x_0)
# p(y=1|x=1)
p_y1_x1 = calculate_posterior(p_x1_y1, p_y_1, p_x_1)
# p(y=0|x=1)
p_y0_x1 = calculate_posterior(p_x1_y0, p_y_0, p_x_1)

print("p(y=1|x=0) = ", p_y1_x0)
print("p(y=0|x=0) = ", p_y0_x0)
print("p(y=1|x=1) = ", p_y1_x1)
print("p(y=0|x=1) = ", p_y0_x1)
```
