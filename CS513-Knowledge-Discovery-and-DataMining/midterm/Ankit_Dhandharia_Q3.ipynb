```python
# Importing all the required Library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

### Part 1

```python
df = pd.read_csv('hepatitis_A.csv')
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

## **Part 2:**

```python
# Defining the columns for ploting
columns = ['BILIRUBIN', 'SGOT', 'ALBUMIN']

# Plotting the scatter plots
for i, col1 in enumerate(columns):
    for col2 in columns[i+1:]:
        plt.figure()
        plt.scatter(df[col1], df[col2], alpha=0.5)
        plt.xlabel(col1.capitalize())
        plt.ylabel(col2.capitalize())
        plt.title(f"Scatter Plot of {col1.capitalize()} vs {col2.capitalize()}")
        plt.show()
```

## **Part 3:**

```python
for col in columns:
    plt.figure()
    df.boxplot(column=col)
    plt.title(f"Box Plot of {col.capitalize()}")
    plt.ylabel(col.capitalize())
    plt.show()
```
