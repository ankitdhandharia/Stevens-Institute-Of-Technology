```python
# Importing the required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## **1. Load the “breast-cancer-wisconsin.data.csv” into Python and perform the EDA analysis**

```python
# Loading the dataset
df = pd.read_csv('drive/MyDrive//breast-cancer-wisconsin.csv')
df.head()
```

### **I. Summarizing each column**

```python
# Desscribing each column
df.describe()
```

### **II. Identifying missing values**

```python
# Checking for missing values
df.isnull().sum()
```

### **III. Replacing the missing values with the “mean” of the column**

```python
# Replacing the missing values with the mean of the column
column_means = df.select_dtypes('number').mean()
for column in column_means.index:
    df[column] = df[column].fillna(column_means[column])
```

### **IV. Displaying the frequency table of “Class” vs. F6**

```python
# Frequency table on Heatmap of Class vs. F6
frequency = pd.crosstab(df['Class'], df['F6'])
plt.figure(figsize=(12, 6))
sns.heatmap(frequency, annot=True, cmap='Blues', fmt='d')
plt.title('Class vs F6')
plt.xlabel('F6')
plt.ylabel('Class')
plt.show()
```

### **V. Displaying the scatter plot of F1 to F6, one pair at a time**

```python
# Displaying the scatter plot of F1 to F6
features = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6']
for i in range(len(features)):
    for j in range(i+1, len(features)):
      plt.scatter(df[features[i]], df[features[j]])
      plt.title(f'{features[i]} vs {features[j]}')
      plt.xlabel(features[i])
      plt.ylabel(features[j])
      plt.show()
```
